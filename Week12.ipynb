Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

Plot 1
Values in subplots takes a bit mroe mental energy to interpret
having the y scale be different with the label mid plot on the first one is both misleading and difficult to read
Plotting as subplots may highlight the differences in a way some people find easier
No legend

Plot 2
Seeing them in the same plot is easier for the reader to interpret, especially with color coding
Since they're in the same plot they also share the same axes values so it is not misleading
Legend would be helpful given color coding
